## Data Quality Audit

### Missing Values

```python
missing_counts = df.isnull().sum()

missing_percentanges = (missing_counts / len(df)) * 100

missing_summary = pd.DataFrame({
    "missing counts" : missing_counts,
    "missing percentages" : missing_percentanges
})

missing_summary
```

### Duplicate Detection

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

### Duplicate Student IDs

```python
duplicate_ids = df["Student_ID"].duplicated().sum()

print("Number of duplicate Student IDs:", duplicate_ids)
```

### Numerical Value Validation

```python
columns_to_check = [
    "Age",
    "Sleep_Hours",
    "Attendance_Rate"
]

for column in columns_to_check:
    print(column)
    print("Min : ", df[column].min())
    print("Max : ", df[column].max())
    print()
```

```python
numerical_columns = df.select_dtypes(include='number')

numerical_columns.agg(['min', 'max']).T
```

### Invalid Numerical Values

```python
invalid_age = (df['Age'] < 0).sum()

invalid_sleep = (
    (df["Sleep_Hours"] < 0) |
    (df["Sleep_Hours"] > 24)
).sum()

invalid_attendance = (
    (df["Attendance_Rate"] < 0) |
    (df["Attendance_Rate"] > 100)
).sum()

print("Invalid Age:", invalid_age)
print("Invalid Sleep Hours:", invalid_sleep)
print("Invalid Attendance Rate:", invalid_attendance)
```

### Potential Outliers — IQR Method

```python
column = "Study_Hours_Per_Week"

q1 = df[column].quantile(0.25)
q3 = df[column].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = (
    (df[column] < lower_bound) |
    (df[column] > upper_bound)
)

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Potential outliers:", outliers.sum())
```

```python
numerical_columns = df.select_dtypes(include="number").columns

for column in numerical_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    print(column, ":", outliers.sum())
```

### Target Validation

```python
allowed_labels = ["Low", "Medium", "High"]

invalid_target = ~df["Mental_Health_Risk"].isin(allowed_labels)

print("Invalid target labels:", invalid_target.sum())
print("Missing target values:", df["Mental_Health_Risk"].isnull().sum())
```

### Data Quality Audit Summary

The dataset contains 2,000 records and 17 columns.

**Main findings:**
- Seven columns contain missing values, with percentages ranging from 1.90% to 4.55%.
- No duplicate rows were detected.
- All Student IDs are unique.
- No invalid values were found in the tested Age, Sleep_Hours, and Attendance_Rate ranges.
- The IQR method identified potential outliers in several numerical features.
- Categorical labels appeared consistent during initial inspection.
- The target variable contains no missing or invalid labels.

**Possible preprocessing strategies:**
- Use median imputation for missing numerical values.
- Use most-frequent imputation for missing categorical values.
- Investigate potential outliers before deciding whether any treatment is necessary.
- Preserve the original dataset and apply fitted preprocessing using training data only.

**Conclusion:**

The dataset is suitable for proceeding to further exploratory analysis and preprocessing planning. No cleaning has been performed during this audit. Dataset-specific feature ranges and potential outliers should be reviewed before making final preprocessing decisions.
